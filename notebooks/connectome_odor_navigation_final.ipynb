## Frozen held-out comparison

```python
results = pd.DataFrame({
    "controller": ["v3 connectome tracking", "v6 tracking + casting"],
    "successes": [70, 71],
    "success_rate": [0.70, 0.71],
    "mean_min_distance": [1.136, 1.385],
    "mean_whiff_fraction": [0.648, 0.616],
    "boundary_contacts": [8, 0],
})
results
```

```python
ax = results.plot.bar(x="controller", y="success_rate", legend=False, ylim=(0,1), rot=0)
ax.set_ylabel("Success rate")
ax.set_title("100 completely unseen random scenarios")
plt.show()
```
